## Analysis

### Univariate analysis

##### Categorical variables

```python
categorical_columns = ['Geography', 'Gender', 'Tenure', 'NumOfProducts', 'HasCrCard',
       'IsActiveMember', 'Exited']

for column in categorical_columns:
    print(f"\n{column}")
    print(df[column].value_counts())

    plt.figure(figsize=(4, 2))
    sns.countplot(data=df, x=column)
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Count")
    plt.show()
```

##### Numerical variables

```python
numerical_columns = [
    'CreditScore',
    'Age',
    'Balance',
    'EstimatedSalary'
]

for column in numerical_columns:
    print(f"\n===== {column} =====")
    print(df[column].describe())

    plt.figure(figsize=(5, 2))
    sns.histplot(data=df, x=column, kde=True)
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.show()

    plt.figure(figsize=(7, 3))
    sns.boxplot(data=df, x=column)
    plt.title(f"Boxplot of {column}")
    plt.xlabel(column)
    plt.show()
```

##### Bivariate analysis

```python
# Bivariate Analysis

categorical_vs_exited = [
    'Geography',
    'Gender',
    'NumOfProducts',
    'IsActiveMember'
]

numerical_vs_exited = [
    'Age',
    'CreditScore',
    'Balance',
    'EstimatedSalary'
]

# Categorical/Discrete vs Exited
for column in categorical_vs_exited:
    plt.figure(figsize=(6, 4))
    sns.countplot(data=df, x=column, hue='Exited')
    plt.title(f'{column} vs Exited')
    plt.xlabel(column)
    plt.ylabel('Count')
    plt.show()


# Numerical vs Exited
for column in numerical_vs_exited:
    plt.figure(figsize=(6, 4))
    sns.boxplot(data=df, x='Exited', y=column)
    plt.title(f'{column} vs Exited')
    plt.xlabel('Exited')
    plt.ylabel(column)
    plt.show()
```

##### Multivariate analysis

```python
features = ['CreditScore', 'Age', 'Balance', 'EstimatedSalary', 'Exited']

sns.pairplot(df[features], hue='Exited')
plt.show()
```

##### Separate features and Target

```python
X = df.drop(['RowNumber', 'CustomerId', 'Surname', 'Exited'], axis = 1)
Y = df['Exited']
```

```python
X.head()
```

##### Encode the Categorical columns

```python
X = pd.get_dummies(X, columns = ['Geography', 'Gender'], dtype = int)
```

```python
X.head()
```

##### Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
SS = StandardScaler()
```

```python
X_Scaled = SS.fit_transform(X)
X_Scaled[:5]
```

##### Train-Test split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X_Scaled, Y, test_size = 0.2, random_state = 42)
```

```python
print(X_train.shape)
print(X_test.shape)
print(Y_train.shape)
print(Y_test.shape)
```

##### Train the Logistic Regression Model

##### Import the Logisctic Regression

```python
from sklearn.linear_model import LogisticRegression
```

##### Create Model

```python
Log_Reg = LogisticRegression()
```

##### Train Model

```python
Log_Reg.fit(X_train, Y_train)
```

##### Make Prdeictions on Test Data

```python
Y_pred = Log_Reg.predict(X_test)
```

```python
Y_pred[:10]
```

##### Check Model accuracy

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(Y_test, Y_pred)
print("Test Accuracy:", accuracy)
```

##### Confusion Matrix

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(Y_test, Y_pred)
print(cm)
```

##### Classification Report

```python
from sklearn.metrics import classification_report
print(classification_report(Y_test, Y_pred))
```

##### Predict on a single row

```python
new_data = X_Scaled[0].reshape(1, -1)
```

```python
prediction = Log_Reg.predict(new_data)
print("Prediction:", prediction)
```
